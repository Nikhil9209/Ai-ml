series:- labelled array that can have multiple data types

Creating series

```python
import numpy as np
import pandas as pd
```

```python
labels =['a','b','c']
my_list=[10,20,30]
arr = np.array([10,20,30])
d ={1:20,2:20,3:30}
```

```python
pd.Series(my_list)
```

```python
 pd.Series(my_list, index=labels)
```

```python
 pd.Series(arr,labels)
```

```python
pd.Series(d)
```
