Pivot table

```python


data = {
    "Salesperson": ["Alice", "Bob", "Alice", "Charlie", "Bob", "Alice"],
    "Region": ["East", "West", "East", "East", "West", "West"],
    "Product": ["Laptop", "Laptop", "Mouse", "Laptop", "Mouse", "Laptop"],
    "Sales": [1200, 1500, 200, 1000, 300, 800]
}

df = pd.DataFrame(data)

print(df)
```

```python
pd.pivot_table(df,values="Sales",index="Region",columns="Product" ,aggfunc='median')
```

Cross tab

```python
pd.crosstab(df['Region'],df['Product'])
```
