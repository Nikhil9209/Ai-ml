```python

import numpy as np

import pandas as pd

employees = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105],
    "Name": ["Rahul", "Aman", "Priya", "Neha", "Karan"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"]
})

print(employees)

```

```python
salary = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 106, 107],
    "Salary": [45000, 55000, 60000, 70000, 50000]
})

print(salary)
```

```python
pd.merge(employees,salary)
```

```python
pd.merge(employees,salary,on="Emp_ID",how="inner" )
```

```python
pd.merge(employees,salary,on="Emp_ID",how="outer" )
```

```python
pd.merge(employees,salary,on="Emp_ID",how="left" )
```

```python
pd.merge(employees,salary,on="Emp_ID",how= "right")
 
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "Emp_ID": [101, 102, 103],
    "Name": ["Rahul", "Aman", "Priya"],
    "Department": ["IT", "HR", "Finance"]
})

print(df1)
```

```python
df2 = pd.DataFrame({
    "Emp_ID": [104, 105, 106],
    "Name": ["Neha", "Karan", "Ravi"],
    "Department": ["IT", "HR", "Finance"]
})

print(df2)
```

```python
result = pd.concat([df1, df2])

print(result)
```

```python
result = pd.concat([df1, df2],axis=1)

print(result)
```

```python
import pandas as pd

employees = pd.DataFrame({
    "Name": ["Rahul", "Aman", "Priya", "Neha"],
    "Department": ["IT", "HR", "Finance", "IT"]
}, index=[101, 102, 103, 104])

print(employees)
```

```python
salary = pd.DataFrame({
    "Salary": [45000, 55000, 60000, 70000],
    "Bonus": [5000, 6000, 7000, 8000]
}, index=[101, 102, 103, 105])

print(salary)
```

```python
result = employees.join(salary)

print(result)
```
